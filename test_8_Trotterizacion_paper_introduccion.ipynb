# II. Introducción

Una evolución unitaria dado un estado inicial $|\psi_{0}\rangle $  se define de la siguiente manera $U(t) = \exp{(-i \, H\, t)}$.  El operador se considera unitario, este se puede simular usando circuitos. El metodo que se utiliza es el conocido como Trotterization donde el hamiltoniano es descompuesto como una suma de la siguiente manera:

$$H = \sum_{i=1}^{k} H_{i}$$

Donde $k$ es el numero de operadores hermiticos que no necesariamente conmutan (i.e. si $A$ y $B$ son operadores estos no cumplen con: $[A,B] = 0$, esto se conoce como conmutación). Se puede aproximar $U$ via:

$$ U(t) ≈ ∏_{j=1}^{n} ∏_{i=1}^{k}\exp{(-i\,Ht / n)}$$

Donde $n \in \mathbb{N}^{+}$ es conocido como la profundidad de trotterization. Entre mas largo es n mejor es la aproximacióin de $U$

### Ejemplo

Como ejemplo vamos a simular una cadena 1D de espines, con una versión simplificada a un hamiltoniano que describe la interaccion entre dos espines vecinos.

$$H = \alpha \, \sigma_x \otimes \sigma_x + \beta \, \sigma_z \otimes \sigma_z $$

### Soluación

Se divide el hamiltoniano en dos partes: $H_1 = \alpha \, \sigma_x \otimes \sigma_x$ y $H_2 = \beta \, \sigma_z \otimes \sigma_z$. En este ejemplo $k=2$. Vamos a usar las siguientes compuertas

$$\begin{split}XX(\phi) = \exp\left(-i \frac{\phi}{2} (X \otimes X)\right) =
\begin{bmatrix} =
    \cos(\phi / 2) & 0 & 0 & -i \sin(\phi / 2) \\
    0 & \cos(\phi / 2) & -i \sin(\phi / 2) & 0 \\
    0 & -i \sin(\phi / 2) & \cos(\phi / 2) & 0 \\
    -i \sin(\phi / 2) & 0 & 0 & \cos(\phi / 2)
\end{bmatrix}.\end{split}$$

y

$$\begin{split}ZZ(\phi) = \exp\left(-i \frac{\phi}{2} (Z \otimes Z)\right) =
\begin{bmatrix}
    e^{-i \phi / 2} & 0 & 0 & 0 \\
    0 & e^{i \phi / 2} & 0 & 0 \\
    0 & 0 & e^{i \phi / 2} & 0 \\
    0 & 0 & 0 & e^{-i \phi / 2}
\end{bmatrix}.\end{split}$$

Para cada una hay ejemplos especiales de la forma:

$$\sigma_x \, \sigma_x (\pi) = i (\sigma_x \otimes \sigma_x)$$

y

$$\sigma_z \, \sigma_z (\pi) = - (\sigma_z \otimes \sigma_z)$$.

## II.A. Conclusion y comentarios finales

### 💡 Resumen: Conmutatividad y Trotterización Cuántica

Sea un Hamiltoniano expresado como la suma de términos más simples:

$$ H = \sum_i H_i $$

---

#### 🔹 Caso 1: Términos conmutativos

Si los operadores $H_i$ **conmutan** entre sí, es decir:

$$ [H_i, H_j] = 0 \quad \forall i,j, $$
entonces la evolución unitaria se factoriza exactamente:

$$ e^{-i H t} = \prod_i e^{-i H_i t} $$

➡️ No hay error de Trotter.  
➡️ Una sola aplicación (depth = 1) reproduce exactamente la evolución unitaria completa.

---

#### 🔹 Caso 2: Términos no conmutativos

Si los operadores **no conmutan**:

$$ [H_i, H_j] \neq 0, $$

la factorización anterior **ya no es exacta**.  
Se usa la **Trotterización (expansión de Trotter-Suzuki)**:

$$ e^{-i H t} =
\lim_{n \to \infty}
\left( \prod_i e^{-i \frac{t}{n} H_i} \right)^n $$

y para un número finito de pasos $n = \text{depth}$:

$$ U_{\text{Trotter}}(t) =
\left( \prod_i e^{-i \frac{t}{n} H_i} \right)^n
\approx e^{-i H t} $$

El error de Trotter es del orden:

$$ \mathcal{O}\left( \frac{t^2}{n} \right) $$

➡️ A mayor profundidad $n$, menor error y mejor aproximación de la evolución exacta.

---

#### 🔹 Resumen conceptual

| Caso | Conmutan los términos | Evolución exacta con depth=1 | Error de Trotter | Acción recomendada |
|------|------------------------|-------------------------------|------------------|--------------------|
| ✅ Conmutativo | Sí | Sí | 0 | Una sola iteración basta |
| ❌ No conmutativo | No | No | $\mathcal{O}(t^2/n)$ | Aumentar `depth` o usar esquemas avanzados |

---

**Interpretación física:**  
- Si los términos del Hamiltoniano conmutan, pueden aplicarse en cualquier orden (o simultáneamente).  
- Si no conmutan, deben intercalarse mediante pasos pequeños (Trotterización) para reproducir correctamente la dinámica cuántica.


# III. Trotterización de Hamiltonianos time dependent

### 💡 Intuición: Trotterización en un Hamiltoniano dependiente del tiempo

There's really not any difference. Imagine I'm trying to simulate a Hamiltonian

$$H(t) = f(t) H_1 + g(t) H_2$$

from time $t = 0$ to $T$.  
I'm going to break this down into $N$ little time steps $\delta = T/N$.  

It's up to you to determine what value of $N$ is large enough to give a reasonable accuracy of simulation.  

So, at each step $n$ (for $t$ between $\delta (n-1)$ and $\delta n$), you're simulating an evolution, approximating it as a constant

$$H(\delta(n - \tfrac{1}{2})),$$

i.e.

$$e^{-i H(\delta(n - \tfrac{1}{2})) \, \delta}.$$

This you will further approximate as

$$
e^{-i H_1 f(\delta(n - \tfrac{1}{2})) \, \delta/2} \;
e^{-i H_2 g(\delta(n - \tfrac{1}{2})) \, \delta} \;
e^{-i H_1 f(\delta(n - \tfrac{1}{2})) \, \delta/2}.
$$

This will build up into a long sequence, just as in the time-independent case,  
except that the different time steps have different weights on the different terms.


```python
import numpy as np
from scipy.linalg import expm
import matplotlib.pyplot as plt
import qutip as qt

# -------------------- Funciones de control dependientes del tiempo --------------------
def f(t):
    """Pulso en el eje X"""
    return np.sin(2 * np.pi * t) + 1.0  # desplazado para ser siempre positivo

def g(t):
    """Pulso en el eje Z"""
    return np.cos(2 * np.pi * t)

# -------------------- Trotterización dependiente del tiempo --------------------
def U_trotterize_time_dependent(T, depth):
    X = np.array([[0, 1], [1, 0]], dtype=complex)
    Z = np.array([[1, 0], [0, -1]], dtype=complex)
    U = np.eye(2, dtype=complex)

    times = np.linspace(0, T, depth)
    dt = T / depth

    for n in range(depth):
        t_mid = (times[n] + times[n-1]) / 2 if n > 0 else dt / 2
        alpha = f(t_mid)
        beta = g(t_mid)
        # Trotter-Suzuki simétrica
        U_step = expm(-1j * alpha * X * dt / 2) @ expm(-1j * beta * Z * dt) @ expm(-1j * alpha * X * dt / 2)
        U = U_step @ U
    return U

# -------------------- Parámetros --------------------
times = np.linspace(0, 10, 1000)
depths = [8, 15, 20]
psi0 = np.array([1, 0], dtype=complex)  # |0>

# -------------------- Evolución Trotterizada --------------------
results_trotter = {d: [] for d in depths}
for d in depths:
    for T in times:
        U = U_trotterize_time_dependent(T, d)
        psi_t = U @ psi0
        results_trotter[d].append(np.abs(psi_t[1])**2)  # prob. de |1>

# -------------------- Evolución exacta (QuTiP) --------------------
X_qt = qt.sigmax()
Z_qt = qt.sigmaz()
psi0_qt = qt.basis(2, 0)

def H_qutip_t(t, args):
    return f(t) * X_qt + g(t) * Z_qt

H_td = [[X_qt, lambda t, args: f(t)], [Z_qt, lambda t, args: g(t)]]
result_exact = qt.mesolve(H_td, psi0_qt, times, [], [])
results_exact = [np.abs(state.full()[1, 0])**2 for state in result_exact.states]

# -------------------- Gráficas --------------------
plt.figure(figsize=(12, 4))
plt.plot(times, f(times), label='$f(t)$ (X)', color='r')
plt.plot(times, g(times), label='$g(t)$ (Z)', color='b')
plt.xlabel('Tiempo')
plt.ylabel('Amplitud del pulso')
plt.title('Perturbaciones sinusoidales en el Hamiltoniano dependiente del tiempo')
plt.legend()
plt.grid(True)
plt.show()

# --- Comparación de evoluciones ---
plt.figure(figsize=(8, 5))
plt.plot(times, results_exact, 'k-', linewidth=2, label='Exacta (QuTiP)')
colors = ['r', 'b', 'g']
for i, d in enumerate(depths):
    plt.plot(times, results_trotter[d], '--', color=colors[i], label=f'Trotter depth={d}')
plt.xlabel('Tiempo')
plt.ylabel('Probabilidad |1>')
plt.title('Evolución con Hamiltoniano dependiente del tiempo: Trotter vs Exacta')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import pennylane as qml
from pennylane import numpy as np

# Parámetros del modelo
N = 5          # número de espines
J = 1.0        # acoplamiento
g = 1.0        # campo transversal
dt = 0.2 / J   # paso de Trotter
n_steps = 5    # número de pasos de Trotter

# Dispositivo: simulador de estado vectorial
dev = qml.device("default.qubit", wires=N)

# --- Construcción del Hamiltoniano ---
def transverse_field_ising(N, J, g):
    terms = []
    coeffs = []
    # Interacciones ZZ
    for j in range(N - 1):
        terms.append(qml.PauliZ(j) @ qml.PauliZ(j + 1))
        coeffs.append(-J)
    # Campo transversal X
    for j in range(N):
        terms.append(qml.PauliX(j))
        coeffs.append(-g)
    return qml.Hamiltonian(coeffs, terms)

H = transverse_field_ising(N, J, g)

# --- Implementación de un paso Trotter simétrico ---
def trotter_step_symmetric(dt):
    # A: campo transversal (X)
    for j in range(N):
        qml.RX(2 * g * dt / 2, wires=j)
    # B: interacciones ZZ (mitad)
    for j in range(0, N - 1, 2):  # pares
        qml.CNOT(wires=[j, j+1])
        qml.RZ(2 * J * dt / 2, wires=j+1)
        qml.CNOT(wires=[j, j+1])
    for j in range(1, N - 1, 2):  # impares
        qml.CNOT(wires=[j, j+1])
        qml.RZ(2 * J * dt, wires=j+1)
        qml.CNOT(wires=[j, j+1])
    for j in range(0, N - 1, 2):  # pares otra vez
        qml.CNOT(wires=[j, j+1])
        qml.RZ(2 * J * dt / 2, wires=j+1)
        qml.CNOT(wires=[j, j+1])
    # A: campo transversal (X) otra vez
    for j in range(N):
        qml.RX(2 * g * dt / 2, wires=j)

# --- Evolución temporal ---
@qml.qnode(dev)
def evolve(n_steps):
    # Estado inicial |↓↓↓↓↓>
    for j in range(N):
        qml.PauliX(j)  # aplica X para obtener |1> en cada qubit

    # Evolución por Trotter
    for _ in range(n_steps):
        trotter_step_symmetric(dt)

    return [qml.expval(qml.PauliZ(j)) for j in range(N)]

# Ejecutar la simulación
result = evolve(n_steps)
print("Magnetización local <Z_j> después de la evolución:", result)
print("Magnetización total:", np.mean(result))
```

```python
# recrear_fig2_pennylane_complete.py
import numpy as np
import pennylane as qml
from pennylane import numpy as pnp
import matplotlib.pyplot as plt
from scipy.linalg import eigh, expm

# -----------------------
# parámetros
# -----------------------
N = 5
J = 1.0
g_list = [1, 2, 3, 4, 5, 6]
dt = 0.2 / J       # paso Trotter según el paper
t_max = 2.0        # eje en J t (0..2)
n_time = 21
times = np.linspace(0.0, t_max, n_time)  # estos son valores de Jt en el paper

dim = 2 ** N
# Paulis
sx = np.array([[0, 1], [1, 0]], dtype=complex)
sz = np.array([[1, 0], [0, -1]], dtype=complex)
si = np.eye(2, dtype=complex)

# -----------------------
# utilidades
# -----------------------
def kron_n(op_list):
    res = op_list[0]
    for op in op_list[1:]:
        res = np.kron(res, op)
    return res

def build_H_matrix(N, J, g):
    H = np.zeros((2**N, 2**N), dtype=complex)
    # ZZ terms
    for j in range(N-1):
        ops = [si] * N
        ops[j] = sz
        ops[j+1] = sz
        H += (-J) * kron_n(ops)
    # X terms
    for j in range(N):
        ops = [si] * N
        ops[j] = sx
        H += (-g) * kron_n(ops)
    return H

def initial_down_state(N):
    v = np.array([0, 1], dtype=complex)  # |1> (spin down in computational basis convention)
    psi = v
    for _ in range(N-1):
        psi = np.kron(psi, v)
    return psi

def local_magnetizations_from_state(psi, N):
    res = np.zeros(N, dtype=float)
    for j in range(N):
        ops = [si] * N
        ops[j] = sz
        Zj = kron_n(ops)
        res[j] = np.vdot(psi, Zj @ psi).real
    return res

# -----------------------
# PennyLane: trotter steps (correct angles según H = -J ZZ - g X)
# -----------------------
dev = qml.device("default.qubit", wires=N)

def trotter_step_first_order(dt, J, g):
    # B_even (pairs 0-1,2-3,...)
    for j in range(0, N-1, 2):
        qml.IsingZZ(-2.0 * J * dt, wires=[j, j+1])
    # B_odd (pairs 1-2,3-4,...)
    for j in range(1, N-1, 2):
        qml.IsingZZ(-2.0 * J * dt, wires=[j, j+1])
    # A: campo X en cada qubit (full step)
    for j in range(N):
        qml.RX(-2.0 * g * dt, wires=j)

def trotter_step_symmetric(dt, J, g):
    # A half
    for j in range(N):
        qml.RX(-g * dt, wires=j)          # RX(angle = -g*dt) -> exp(i g X dt/2)
    # B_odd half (pairs 1-2, 3-4, ...)
    for j in range(1, N-1, 2):
        qml.IsingZZ(-1.0 * J * dt, wires=[j, j+1])   # half step angle
    # C_even full (pairs 0-1,2-3,...)
    for j in range(0, N-1, 2):
        qml.IsingZZ(-2.0 * J * dt, wires=[j, j+1])   # full step angle
    # B_odd half again
    for j in range(1, N-1, 2):
        qml.IsingZZ(-1.0 * J * dt, wires=[j, j+1])
    # A half final
    for j in range(N):
        qml.RX(-g * dt, wires=j)

# qnode que acepta g como parámetro y el esquema
@qml.qnode(dev)
def qnode_trotter(n_steps: int, scheme: str, g: float):
    # preparar |↓↓...↓> = aplicar X en cada qubit sobre |0>
    for j in range(N):
        qml.PauliX(wires=j)
    # aplicar n_steps de Trotter
    for _ in range(n_steps):
        if scheme == "first":
            trotter_step_first_order(dt, J, g)
        elif scheme == "symmetric":
            trotter_step_symmetric(dt, J, g)
        else:
            raise ValueError("scheme must be 'first' or 'symmetric'")
    # medir <Z_j> para cada qubit
    return [qml.expval(qml.PauliZ(j)) for j in range(N)]

# -----------------------
# Simulación y plotting: layout 4x3 como en Fig.2
# -----------------------
psi0 = initial_down_state(N)

fig, axes = plt.subplots(4, 3, figsize=(12, 12), sharex=True, sharey=True)
plt.subplots_adjust(hspace=0.35, wspace=0.25)
axes = axes.flatten()

for ig, gval in enumerate(g_list):
    # reconstruir Hamiltoniano exacto para este g
    H_exact = build_H_matrix(N, J, gval)

    M_exact = []
    M_first = []
    M_symm = []

    for t_Jt in times:
        # times are in units J*t (paper axis). Convert to physical time t_phys = (Jt)/J
        t_phys = t_Jt / J
        # evolución exacta U = exp(-i H t_phys)
        U_t = expm(-1j * H_exact * t_phys)
        psi_t = U_t @ psi0
        local_exact = local_magnetizations_from_state(psi_t, N)
        M_exact.append(np.mean(local_exact))

        # calcular número de pasos de Trotter: n_steps ~ t_phys / dt
        # pero dt = 0.2 / J, y t_phys = t_Jt / J => n_steps = round(t_Jt / 0.2) (independiente de J)
        n_steps = int(np.round(t_Jt / 0.2))
        # asegurarse n_steps >= 0
        if n_steps < 0:
            n_steps = 0

        # trotter (qnode) - ideal (statevector)
        vals_first = qnode_trotter(n_steps, "first", gval)
        vals_symm = qnode_trotter(n_steps, "symmetric", gval)

        M_first.append(np.mean([float(v) for v in vals_first]))
        M_symm.append(np.mean([float(v) for v in vals_symm]))

    # plot: top 2 filas -> first-order (panels 0..5), bottom 2 filas -> symmetric (panels 6..11)
    ax_first = axes[ig]            # 0..5
    ax_symm = axes[6 + ig]        # 6..11

    # First-order panel
    ax_first.plot(times, M_exact, color="k", lw=1.5, label="exact")
    ax_first.plot(times, M_first, marker='o', markersize=4, label="ideal (first-order)")
    ax_first.set_title(f"first-order, g={gval}J")
    ax_first.set_ylim([-1.05, 1.05])
    ax_first.set_xlim([0, t_max])
    if ig % 3 == 0:
        ax_first.set_ylabel("M")
    if ig >= 3 and ig % 3 == 0:
        ax_first.set_ylabel("M")

    # Symmetric panel
    ax_symm.plot(times, M_exact, color="k", lw=1.5, label="exact")
    ax_symm.plot(times, M_symm, marker='s', markersize=4, label="ideal (symmetric)")
    ax_symm.set_title(f"symmetric, g={gval}J")
    ax_symm.set_ylim([-1.05, 1.05])
    ax_symm.set_xlim([0, t_max])
    if (6 + ig) % 3 == 0:
        ax_symm.set_ylabel("M")
    if (6 + ig) >= 9:
        ax_symm.set_xlabel("J t")

# Leyenda global (tomada del primer subplot)
handles, labels = axes[0].get_legend_handles_labels()
fig.legend(handles, labels, loc='upper center', ncol=3, bbox_to_anchor=(0.5, 0.99))

plt.suptitle("Fig.2-like: Exact vs First-order (top) and Exact vs Symmetric (bottom)\nN=5, dt=0.2/J (ideal simulator)", y=0.995)
plt.show()
```



## 1) Objetivo general

El código simula la **dinámica temporal** de una cadena de $N=4$ espines interactuantes bajo un Hamiltoniano tipo Heisenberg/Zeeman con **acoplamientos dependientes del tiempo** $J_{ij}(t)$ y **pulsos magnéticos dependientes del tiempo** $\mathbf{B}_i(t)$. La simulación se hace mediante **Trotterización de primer orden**, es decir, se divide la evolución total en pequeños pasos $\Delta t$ y en cada paso se aplica aproximadamente la evolución debida a las interacciones y al campo local (suponiendo que son constantes durante $\Delta t$).

Al final se grafica:

* Los pulsos en el tiempo (el perfil de $J(t)$ y las componentes $B_z^{(i)}(t)$ por sitio)
* La magnetización total $M(t)=\frac{1}{N}\sum_j \langle Z_j\rangle$

---

## 2) Parámetros y dispositivo

* `N = 4`: número de qubits/espines.
* `total_time = 2.0` y `n_trotter_steps = 80` ⇒ `dt = total_time / n_trotter_steps` es el tamaño de paso Trotter.
* `dev = qml.device("default.qubit", wires=N)`: usa el simulador de estado vectorial (no es un hardware real).

---

## 3) Pulsos cuadrados

* `square_pulse(t, intervals)` toma una lista de intervalos `(t_start, t_end, amplitude)` y devuelve la amplitud activa en el tiempo `t` (suma pulsos si se superponen).
* `J_pulse_intervals` define dos pulsos rectangulares para el acoplamiento global $J(t)$.
* `B_pulse_intervals_per_site` define para cada sitio una lista de pulsos vectoriales $(B_x,B_y,B_z)$ activos en distintos intervalos.

Estas funciones permiten modelar señales que “se encienden” y “se apagan” en tiempos concretos.

---

## 4) Cómo se Trotteriza el Hamiltoniano

El término de interacción entre dos spins vecinos $S_i\cdot S_j$ se escribe en términos de matrices de Pauli como

$$
S_i\cdot S_j = \tfrac{1}{4}\big(\sigma_x\sigma_x+\sigma_y\sigma_y+\sigma_z\sigma_z\big)
$$

(porque $S=\tfrac12\sigma$).
En PennyLane hay puertas `IsingXX`, `IsingYY`, `IsingZZ` que implementan exponentes tipo $e^{-i (\theta/2) \sigma_a\otimes\sigma_a}$. Para reproducir $e^{-iJ_{ij}\,dt\,(S_i\cdot S_j)}$ se escoge el ángulo
$\theta = J_{ij}\,dt/2$ y se aplica la secuencia `IsingXX(theta)`, `IsingYY(theta)`, `IsingZZ(theta)`.
Eso es lo que hace la función `apply_two_body_interaction(Jij, dt, i, j)`.

Para el término Zeeman local $-g_L\mu_B \mathbf{B}_i\cdot \mathbf{S}_i$ (recordando $S_i=(1/2)\sigma$), una rotación de Pauli `RX`, `RY`, `RZ` con ángulo apropiado implementa la evolución local. La función `apply_local_zeeman(Bvec, dt, i)` aplica rotaciones `RX/RY/RZ` con ángulo ( $-g\_L \mu\_B B\_{comp} \cdot dt\$ ` para cada componente no nula del campo.

---

## 5) Preparación del estado y bucle Trotter

* En el `qnode` `trotter_evolution(n_steps)`, primero se prepara el estado inicial: **todos los qubits en el estado "down"** usando `qml.PauliX` sobre $|0\rangle$ (esto hace que la proyección $ \langle Z\rangle $ inicial sea $-1$ por qubit).
* Luego se hace un bucle de `n_steps` pasos. En cada paso:

  * se evalúa el tiempo representativo del paso `t = (k + 0.5) * dt` (se usa el punto medio del subintervalo para mejor precisión).
  * se recorre cada enlace `(i,j)` y, si $J_{ij}(t)$ no es cero, se aplica la interacción entre esos dos qubits con `IsingXX/YY/ZZ`.
  * después se aplica el término Zeeman local en cada sitio con rotaciones `RX/RY/RZ` según el valor de $\mathbf{B}_i(t)$.
* Finalmente el `qnode` devuelve las esperanzas $\langle Z_i\rangle$ en cada qubit.

---

## 6) Muestreo temporal y cálculo de la magnetización

* Fuera del `qnode`, el script recorre `k = 0..n_trotter_steps` y llama al `qnode` con `n_steps = k` para obtener la evolución en cada tiempo discretizado.

  * `k=0` usa el valor inicial `M=-1`.
  * Para cada `k>0` se llama al `qnode` y se computa la magnetización total media `M_tot = mean(<Z_i>)`.
* **Nota eficiente**: este enfoque hace muchas llamadas al qnode (una por cada tiempo). En simulador local no es grave; en backends remotos esto sería ineficiente (mejor sería que el qnode devolviera todas las expectativas intermedias en una sola ejecución o agrupar trabajos).

---

## 7) Pre-cálculo de los pulsos para graficar

Antes de ejecutar la evolución se muestrean los pulsos en la malla `times`:

* `J_vals` contiene $J(t)$ para cada tiempo.
* `Bz_vals`, `Bx_vals`, `By_vals` contienen las componentes del campo por sitio en cada tiempo.
  Esto permite superponer las señales y la magnetización en la misma figura.

---

## 8) Gráficas finales

Se crean dos subplots apilados:

* Panel superior: muestra `J(t)` (línea gris continua) y `B_z` por sitio (líneas discontinuas con colores distintos).
* Panel inferior: muestra la magnetización total `M(t)` a los tiempos muestreados.

Así obtienes visualmente cómo los pulsos (activación de acoplamientos y campos) influyen en la dinámica de la magnetización.

---

## 9) Limites, precisión y recomendaciones prácticas

* **Error de Trotter:** la aproximación es válida para `dt` pequeño. Si tus pulsos tienen bordes muy rapidos, incrementa `n_trotter_steps` para reducir el error.
* **Profundidad de circuito:** cada paso aplica muchas puertas (IsingXX/YY/ZZ + rotaciones locales). Al aumentar `n_steps` el circuito se hace largo; en hardware real puede ser inviable por ruido.
* **Eficiencia:** llamar al `qnode` para cada tiempo es costoso si usas un backend remoto (Braket/IBM). Para ejecución remota conviene agrupar tiempos o construir un único circuito que devuelva estados intermedios (o usar simulador local).
* **Extensiones fáciles:**

  * Trotter simétrica (2º orden) para mejorar error Trotter.
  * Calcular energía esperada ⟨H(t)⟩ construyendo la matriz H(t) y evaluando la expectativa si quieres verificar conservación/escalas.
  * Graficar componentes Bx/By o la norma del campo si te interesa la magnitud total.

---


```python
# trotter_time_dependent_heisenberg_with_pulses.py
import numpy as np
import pennylane as qml
import matplotlib.pyplot as plt

# -----------------------
# Parámetros físicos & sistema
# -----------------------
N = 4                    # número de spins
J_base = 1.0             # escala de acoplamiento (si lo deseas)
g_L = 1.0                # Landé g-factor (escala)
mu_B = 1.0               # Bohr magneton (escala) -> puedes poner valor físico
total_time = 2.0         # tiempo total de evolución (unidades arbitrarias)
n_trotter_steps = 80     # número de pasos Trotter (dt = total_time / n_trotter_steps)
dt = total_time / n_trotter_steps

# dispositivo PennyLane (simulador statevector)
dev = qml.device("default.qubit", wires=N)

# -----------------------
# Pulsos cuadrados: utilidad para definir señales en el tiempo
# -----------------------
def square_pulse(t, intervals):
    """
    intervals: lista de (t_start, t_end, amplitude)
    devuelve la suma de amplitudes activas en tiempo t
    """
    val = 0.0
    for (t0, t1, amp) in intervals:
        if (t >= t0) and (t < t1):
            val += amp
    return val

# -----------------------
# Definición de pulsos (ejemplo)
# -----------------------
J_pulse_intervals = [
    (0.3, 0.8, 2.0),
    (1.2, 1.6, -1.0)
]

B_pulse_intervals_per_site = [
    [(0.0, 0.5, (0.0, 0.0, 0.5)), (1.0, 1.5, (0.2, 0.0, 0.0))],  # site 0
    [(0.2, 0.9, (0.0, 0.0, 0.4))],                              # site 1
    [(0.4, 1.2, (0.0, 0.1, 0.0))],                              # site 2
    [(0.6, 1.8, (0.0, 0.0, 0.6))]                               # site 3
]

edge_list = [(i, i+1) for i in range(N-1)]

def J_ij_at_time(i, j, t):
    amp = square_pulse(t, J_pulse_intervals)
    return J_base * amp

def B_i_at_time(i, t):
    intervals = B_pulse_intervals_per_site[i] if i < len(B_pulse_intervals_per_site) else []
    Bx = 0.0; By = 0.0; Bz = 0.0
    for (t0, t1, vec) in intervals:
        if (t >= t0) and (t < t1):
            bx, by, bz = vec
            Bx += bx; By += by; Bz += bz
    return np.array([Bx, By, Bz], dtype=float)

# -----------------------
# Implementación de un paso Trotter (primer orden)
# -----------------------
def apply_two_body_interaction(Jij, dt, i, j):
    theta = Jij * dt / 2.0
    qml.IsingXX(theta, wires=[i, j])
    qml.IsingYY(theta, wires=[i, j])
    qml.IsingZZ(theta, wires=[i, j])

def apply_local_zeeman(Bvec, dt, i, gL=g_L, muB=mu_B):
    Bx, By, Bz = Bvec
    if abs(Bx) > 0:
        qml.RX(-gL * muB * Bx * dt, wires=i)
    if abs(By) > 0:
        qml.RY(-gL * muB * By * dt, wires=i)
    if abs(Bz) > 0:
        qml.RZ(-gL * muB * Bz * dt, wires=i)

# -----------------------
# QNode que realiza una evolución Trotterizada hasta tiempo T (usando n_steps)
# -----------------------
@qml.qnode(dev)
def trotter_evolution(n_steps, scheme="first"):
    # Estado inicial: todos down |1> (aplicamos PauliX sobre |0>)
    for w in range(N):
        qml.PauliX(wires=w)

    # loop Trotter
    for k in range(n_steps):
        t = (k + 0.5) * dt
        # interacciones
        for (i, j) in edge_list:
            Jij = J_ij_at_time(i, j, t)
            if abs(Jij) < 1e-12:
                continue
            apply_two_body_interaction(Jij, dt, i, j)
        # Zeeman local
        for i in range(N):
            Bvec = B_i_at_time(i, t)
            if np.allclose(Bvec, 0.0):
                continue
            apply_local_zeeman(Bvec, dt, i)

    return [qml.expval(qml.PauliZ(i)) for i in range(N)]

# -----------------------
# Simulación: calcular M(t) y también muestrear pulsos en el mismo grid temporal
# -----------------------
times = np.linspace(0.0, total_time, n_trotter_steps + 1)  # incluye t=0
M_t = []

# Pre-computar pulsos en los puntos `times` para graficar
J_vals = np.array([J_ij_at_time(0, 1, t) for t in times])  # J global (misma en todos los enlaces en el ejemplo)
# Para B, extraemos la componente z (puedes cambiar a x,y o norma)
Bz_vals = np.zeros((N, len(times)))
Bx_vals = np.zeros((N, len(times)))
By_vals = np.zeros((N, len(times)))
for idx, t in enumerate(times):
    for i in range(N):
        Bvec = B_i_at_time(i, t)
        Bx_vals[i, idx] = Bvec[0]
        By_vals[i, idx] = Bvec[1]
        Bz_vals[i, idx] = Bvec[2]

# Ejecutar evolución Trotter para cada n_steps (discreto)
for k in range(n_trotter_steps + 1):
    if k == 0:
        M_t.append(-1.0)  # estado inicial todos down
    else:
        vals = trotter_evolution(k, scheme="first")
        M_tot = np.mean([float(v) for v in vals])
        M_t.append(M_tot)

M_t = np.array(M_t)

# -----------------------
# Graficar: pulsos vs magnetización
# -----------------------
fig, (ax_pulses, ax_mag) = plt.subplots(2, 1, figsize=(10, 7), sharex=True,
                                         gridspec_kw={'height_ratios': [1, 1.1]})

# Pulsos: J(t)
ax_pulses.plot(times, J_vals, label='J(t) (global)', color='tab:gray', linewidth=2)

# Pulsos: Bz por sitio (usar diferentes colores)
colors = plt.cm.viridis(np.linspace(0, 1, N))
for i in range(N):
    ax_pulses.plot(times, Bz_vals[i], '--', label=f'Bz site {i}', color=colors[i])

ax_pulses.set_ylabel('Pulse amplitude')
ax_pulses.set_title('Pulsos en tiempo: J(t) y B_z^{(i)}(t)')
ax_pulses.legend(loc='upper right', fontsize='small', ncol=2)
ax_pulses.grid(True)

# Magnetización
ax_mag.plot(times, M_t, marker='o', linestyle='-', color='tab:red', label='M(t)')
ax_mag.set_xlabel('t')
ax_mag.set_ylabel('M(t) = (1/N) Σ ⟨Z_j⟩')
ax_mag.set_title('Magnetización total')
ax_mag.grid(True)
ax_mag.legend()

plt.tight_layout()
plt.show()
```

```python
# trotter_vs_exact_time_dependent.py
import numpy as np
import pennylane as qml
import matplotlib.pyplot as plt
from scipy.linalg import expm

# -----------------------
# Parámetros físicos & sistema
# -----------------------
N = 4                    # número de spins
J_base = 1.0             # escala de acoplamiento
g_L = 1.0                # Landé g-factor
mu_B = 1.0               # Bohr magneton (escala)
total_time = 12.0
n_trotter_steps = 80     # número de pasos Trotter para ambas integraciones
dt = total_time / n_trotter_steps

# dispositivo PennyLane (simulador statevector) para trotter
dev = qml.device("default.qubit", wires=N)

# -----------------------
# Pulsos cuadrados: utilidad para definir señales en el tiempo
# -----------------------
def square_pulse(t, intervals):
    """
    intervals: lista de (t_start, t_end, amplitude)
    devuelve la suma de amplitudes activas en tiempo t
    """
    val = 0.0
    for (t0, t1, amp) in intervals:
        if (t >= t0) and (t < t1):
            val += amp
    return val

# -----------------------
# Definición de pulsos (ejemplo)
# -----------------------
# --- J(t): apagado (sin interacción entre spins)
J_pulse_intervals = [
    (0.0, 12.0, 0.0)
]

# --- Campos locales: pulso en X positivo y luego negativo en todos los sitios
B_pulse_intervals_per_site = [
    [(2.0, 2.4, (3, 0.0, 0.0)),   # campo Bx positivo
     (7.0, 10.0, (-4.0, 0.0, 0.0))],  # campo Bx negativo
] * N

edge_list = [(i, i+1) for i in range(N-1)]

def J_ij_at_time(i, j, t):
    amp = square_pulse(t, J_pulse_intervals)
    return J_base * amp

def B_i_at_time(i, t):
    intervals = B_pulse_intervals_per_site[i] if i < len(B_pulse_intervals_per_site) else []
    Bx = 0.0; By = 0.0; Bz = 0.0
    for (t0, t1, vec) in intervals:
        if (t >= t0) and (t < t1):
            bx, by, bz = vec
            Bx += bx; By += by; Bz += bz
    return np.array([Bx, By, Bz], dtype=float)

# -----------------------
# Utilidades para construir Hamiltoniano matricial (exacta)
# -----------------------
# Pauli matrices
sx = np.array([[0, 1], [1, 0]], dtype=complex)
sy = np.array([[0, -1j], [1j, 0]], dtype=complex)
sz = np.array([[1, 0], [0, -1]], dtype=complex)
si = np.eye(2, dtype=complex)

def kron_n(op_list):
    out = op_list[0]
    for op in op_list[1:]:
        out = np.kron(out, op)
    return out

def build_H_matrix_time(t):
    """
    Construye la matriz del Hamiltoniano H(t) = sum_{<i,j>} J_ij(t) S_i·S_j - gL muB sum_i B_i(t)·S_i
    con S = 1/2 sigma.
    """
    dim = 2**N
    H = np.zeros((dim, dim), dtype=complex)

    # término interacción S_i·S_j = 1/4 (σxσx + σyσy + σzσz)
    for (i, j) in edge_list:
        Jij = J_ij_at_time(i, j, t)
        if abs(Jij) < 1e-12:
            continue
        for (pa, op) in [('x', sx), ('y', sy), ('z', sz)]:
            ops = [si] * N
            ops[i] = op
            ops[j] = op
            H += (-Jij * 0.25) * kron_n(ops)  # note sign convention: H_int = sum (-J S·S) ?

    # término Zeeman local: - g_L mu_B B_i · S_i  with S_i = sigma/2
    for i in range(N):
        Bx, By, Bz = B_i_at_time(i, t)
        if abs(Bx) > 1e-12:
            ops = [si] * N
            ops[i] = sx
            H += (-g_L * mu_B * Bx * 0.5) * kron_n(ops)
        if abs(By) > 1e-12:
            ops = [si] * N
            ops[i] = sy
            H += (-g_L * mu_B * By * 0.5) * kron_n(ops)
        if abs(Bz) > 1e-12:
            ops = [si] * N
            ops[i] = sz
            H += (-g_L * mu_B * Bz * 0.5) * kron_n(ops)

    return H

# -----------------------
# Implementación de Trotter (primer orden) en PennyLane (como antes)
# -----------------------
def apply_two_body_interaction(Jij, dt, i, j):
    theta = Jij * dt / 2.0
    qml.IsingXX(theta, wires=[i, j])
    qml.IsingYY(theta, wires=[i, j])
    qml.IsingZZ(theta, wires=[i, j])

def apply_local_zeeman(Bvec, dt, i, gL=g_L, muB=mu_B):
    Bx, By, Bz = Bvec
    if abs(Bx) > 0:
        qml.RX(-gL * muB * Bx * dt, wires=i)
    if abs(By) > 0:
        qml.RY(-gL * muB * By * dt, wires=i)
    if abs(Bz) > 0:
        qml.RZ(-gL * muB * Bz * dt, wires=i)

@qml.qnode(dev)
def trotter_evolution(n_steps, scheme="first"):
    # Estado inicial: todos down |1>
    for w in range(N):
        qml.PauliX(wires=w)

    for k in range(n_steps):
        t = (k + 0.5) * dt
        # interacciones
        for (i, j) in edge_list:
            Jij = J_ij_at_time(i, j, t)
            if abs(Jij) < 1e-12:
                continue
            apply_two_body_interaction(Jij, dt, i, j)
        # Zeeman local
        for i in range(N):
            Bvec = B_i_at_time(i, t)
            if np.allclose(Bvec, 0.0):
                continue
            apply_local_zeeman(Bvec, dt, i)

    return [qml.expval(qml.PauliZ(i)) for i in range(N)]

# -----------------------
# Estado inicial (vector) para evolución exacta
# -----------------------
def initial_down_state(N):
    v = np.array([0,1], dtype=complex)   # |1> per qubit (spin down)
    psi = v
    for _ in range(N-1):
        psi = np.kron(psi, v)
    return psi

def local_magnetizations_from_state(psi):
    """Devuelve vector de <Z_j> para el estado psi (matriz completa)."""
    res = np.zeros(N, dtype=float)
    dim = 2**N
    for j in range(N):
        ops = [np.eye(2, dtype=complex)] * N
        ops[j] = sz
        Zj = kron_n(ops)
        res[j] = np.vdot(psi, Zj @ psi).real
    return res

# -----------------------
# Integración exacta en pasos (time-ordered exponential por pasos)
# -----------------------
times = np.linspace(0.0, total_time, n_trotter_steps + 1)  # incluye t=0
M_t_trotter = []   # magnetización trotter (ya en tu workflow)
M_t_exact = []     # magnetización exacta via expm stepping

# calcular pulsos para plotting (igual que antes)
J_vals = np.array([J_ij_at_time(0, 1, t) for t in times])
#Bz_vals = np.zeros((N, len(times)))
#for idx, t in enumerate(times):
#    for i in range(N):
#        Bvec = B_i_at_time(i, t)
#        Bz_vals[i, idx] = Bvec[2]

Bx_vals = np.zeros((N, len(times)))
Bz_vals = np.zeros((N, len(times)))
for idx, t in enumerate(times):
    for i in range(N):
        Bvec = B_i_at_time(i, t)
        Bx_vals[i, idx] = Bvec[0]
        Bz_vals[i, idx] = Bvec[2]

# --- Evolución exacta por pasos (matriz) ---
psi_exact = initial_down_state(N)
M_t_exact.append(np.mean(local_magnetizations_from_state(psi_exact)))  # t=0

# --- Simulación trotter (usaremos qnode calls like before) ---
# Nota: trotter_evolution(k) devuelve <Z_j> al final de k pasos
M_t_trotter.append(np.mean(local_magnetizations_from_state(psi_exact)))  # t=0

print("Comenzando bucle temporal: se llamará al qnode trotter para cada k y se integrará la evolución exacta por pasos")

for k in range(1, n_trotter_steps + 1):
    # tiempo del paso k (midpoint t for building H)
    t_mid = (k - 0.5) * dt

    # --- Exact propagation: U = exp(-i H(t_mid) * dt) applied to psi_exact
    H_mid = build_H_matrix_time(t_mid)
    U = expm(-1j * H_mid * dt)
    psi_exact = U @ psi_exact
    mz_exact = np.mean(local_magnetizations_from_state(psi_exact))
    M_t_exact.append(mz_exact)

    # --- Trotter result via PennyLane qnode
    vals = trotter_evolution(k, scheme="first")
    mz_trot = np.mean([float(v) for v in vals])
    M_t_trotter.append(mz_trot)

    if (k % 10) == 0:
        print(f" step {k}/{n_trotter_steps}: t={(k*dt):.3f}  M_exact={mz_exact:.4f}  M_trot={mz_trot:.4f}")

# convertir a arrays
M_t_exact = np.array(M_t_exact)
M_t_trotter = np.array(M_t_trotter)

# -----------------------
# Graficar: pulsos vs magnetización comparación
# -----------------------
fig, (ax_pulses, ax_mag) = plt.subplots(2, 1, figsize=(10, 7), sharex=True,
                                         gridspec_kw={'height_ratios': [1, 1.1]})

# Pulsos: J(t)
ax_pulses.plot(times, J_vals, label='J(t) (global)', color='tab:gray', linewidth=2)

# Pulsos: Bz por sitio
colors = plt.cm.viridis(np.linspace(0, 1, N))
#for i in range(N):
#    ax_pulses.plot(times, Bz_vals[i], '--', label=f'Bz site {i}', color=colors[i])

for i in range(N):
    ax_pulses.plot(times, Bx_vals[i], '--', label=f'Bx site {i}', color=colors[i])

ax_pulses.set_ylabel('Pulse amplitude')
ax_pulses.set_title('Pulsos en tiempo: J(t) y B_z^{(i)}(t)')
ax_pulses.legend(loc='upper right', fontsize='small', ncol=2)
ax_pulses.grid(True)

# Magnetización: comparar exact vs trotter
ax_mag.plot(times, M_t_exact, '-', color='k', lw=1.5, label='exact (matrix exp stepping)')
ax_mag.plot(times, M_t_trotter, 'o--', color='tab:red', label='trotter (PennyLane)')
ax_mag.set_xlabel('t')
ax_mag.set_ylabel('M(t) = (1/N) Σ ⟨Z_j⟩')
ax_mag.set_title('Magnetización total (Exact vs Trotter)')
ax_mag.grid(True)
ax_mag.legend()

plt.tight_layout()
plt.show()
```

```python
# trotter_vs_exact_time_dependent_fixed.py
import numpy as np
import pennylane as qml
import matplotlib.pyplot as plt
from scipy.linalg import expm

# -----------------------
# Parámetros físicos & sistema
# -----------------------
N = 4                    # número de spins
J_base = 1.0             # escala de acoplamiento
g_L = 1.0                # Landé g-factor
mu_B = 1.0               # Bohr magneton (escala)
total_time = 12.0
n_trotter_steps = 80     # número de pasos Trotter (coarse)
# dt_trot se calcula más abajo desde n_trotter_steps

# dispositivo PennyLane (simulador statevector) para trotter
dev = qml.device("default.qubit", wires=N)

# -----------------------
# Pulsos cuadrados: utilidad para definir señales en el tiempo
# -----------------------
def square_pulse(t, intervals):
    """
    intervals: lista de (t_start, t_end, amplitude)
    devuelve la suma de amplitudes activas en tiempo t
    """
    val = 0.0
    for (t0, t1, amp) in intervals:
        if (t >= t0) and (t < t1):
            val += amp
    return val

# -----------------------
# Definición de pulsos (ejemplo)
# -----------------------
# --- J(t): apagado (sin interacción entre spins)
J_pulse_intervals = [
    (0.0, 12.0, 0.0)
]

# --- Campos locales: pulso en X positivo y luego negativo en todos los sitios
B_pulse_intervals_per_site = [
    [(2.0, 2.4, (3.0, 0.0, 0.0)),   # campo Bx positivo corto
     (7.0, 10.0, (-4.0, 0.0, 0.0))],  # campo Bx negativo más largo
] * N

edge_list = [(i, i+1) for i in range(N-1)]

def J_ij_at_time(i, j, t):
    amp = square_pulse(t, J_pulse_intervals)
    return J_base * amp

def B_i_at_time(i, t):
    intervals = B_pulse_intervals_per_site[i] if i < len(B_pulse_intervals_per_site) else []
    Bx = 0.0; By = 0.0; Bz = 0.0
    for (t0, t1, vec) in intervals:
        if (t >= t0) and (t < t1):
            bx, by, bz = vec
            Bx += bx; By += by; Bz += bz
    return np.array([Bx, By, Bz], dtype=float)

# -----------------------
# Utilidades para construir Hamiltoniano matricial (exacta)
# -----------------------
# Pauli matrices
sx = np.array([[0, 1], [1, 0]], dtype=complex)
sy = np.array([[0, -1j], [1j, 0]], dtype=complex)
sz = np.array([[1, 0], [0, -1]], dtype=complex)
si = np.eye(2, dtype=complex)

def kron_n(op_list):
    out = op_list[0]
    for op in op_list[1:]:
        out = np.kron(out, op)
    return out

def build_H_matrix_time(t):
    """
    Construye la matriz del Hamiltoniano H(t) = sum_{<i,j>} J_ij(t) S_i·S_j - gL muB sum_i B_i(t)·S_i
    con S = 1/2 sigma.
    """
    dim = 2**N
    H = np.zeros((dim, dim), dtype=complex)

    # término interacción S_i·S_j = 1/4 (σxσx + σyσy + σzσz)
    for (i, j) in edge_list:
        Jij = J_ij_at_time(i, j, t)
        if abs(Jij) < 1e-12:
            continue
        for op in (sx, sy, sz):
            ops = [si] * N
            ops[i] = op
            ops[j] = op
            H += (-Jij * 0.25) * kron_n(ops)

    # término Zeeman local: - g_L mu_B B_i · S_i  with S_i = sigma/2
    for i in range(N):
        Bx, By, Bz = B_i_at_time(i, t)
        if abs(Bx) > 1e-12:
            ops = [si] * N
            ops[i] = sx
            H += (-g_L * mu_B * Bx * 0.5) * kron_n(ops)
        if abs(By) > 1e-12:
            ops = [si] * N
            ops[i] = sy
            H += (-g_L * mu_B * By * 0.5) * kron_n(ops)
        if abs(Bz) > 1e-12:
            ops = [si] * N
            ops[i] = sz
            H += (-g_L * mu_B * Bz * 0.5) * kron_n(ops)

    return H

# -----------------------
# Implementación de Trotter (primer orden) en PennyLane
# -----------------------
def apply_two_body_interaction(Jij, dt_local, i, j):
    theta = Jij * dt_local / 2.0
    qml.IsingXX(theta, wires=[i, j])
    qml.IsingYY(theta, wires=[i, j])
    qml.IsingZZ(theta, wires=[i, j])

def apply_local_zeeman(Bvec, dt_local, i, gL=g_L, muB=mu_B):
    Bx, By, Bz = Bvec
    if abs(Bx) > 0:
        qml.RX(-gL * muB * Bx * dt_local, wires=i)
    if abs(By) > 0:
        qml.RY(-gL * muB * By * dt_local, wires=i)
    if abs(Bz) > 0:
        qml.RZ(-gL * muB * Bz * dt_local, wires=i)

@qml.qnode(dev)
def trotter_evolution(n_steps, scheme="first"):
    # Estado inicial: todos down |1>
    for w in range(N):
        qml.PauliX(wires=w)

    dt_trot_local = total_time / n_trotter_steps
    for k in range(n_steps):
        t = (k + 0.5) * dt_trot_local
        # interacciones
        for (i, j) in edge_list:
            Jij = J_ij_at_time(i, j, t)
            if abs(Jij) < 1e-12:
                continue
            apply_two_body_interaction(Jij, dt_trot_local, i, j)
        # Zeeman local
        for i in range(N):
            Bvec = B_i_at_time(i, t)
            if np.allclose(Bvec, 0.0):
                continue
            apply_local_zeeman(Bvec, dt_trot_local, i)

    return [qml.expval(qml.PauliZ(i)) for i in range(N)]

# -----------------------
# Estado inicial (vector) para evolución exacta
# -----------------------
def initial_down_state(N):
    v = np.array([0,1], dtype=complex)   # |1> per qubit (spin down)
    psi = v
    for _ in range(N-1):
        psi = np.kron(psi, v)
    return psi

def local_magnetizations_from_state(psi):
    """Devuelve vector de <Z_j> para el estado psi (matriz completa)."""
    res = np.zeros(N, dtype=float)
    for j in range(N):
        ops = [np.eye(2, dtype=complex)] * N
        ops[j] = sz
        Zj = kron_n(ops)
        res[j] = np.vdot(psi, Zj @ psi).real
    return res

# -----------------------
# Integración exacta en pasos (time-ordered exponential por pasos)
# -----------------------
n_exact_substeps = 100             # subpasos por paso trotter (ajustable)
n_exact_steps = n_trotter_steps * n_exact_substeps
dt_exact = total_time / n_exact_steps
dt_trot = total_time / n_trotter_steps

# tiempos a muestrear para graficar (coinciden con los pasos Trotter)
times_trot = np.linspace(0.0, total_time, n_trotter_steps + 1)

times_exact = np.linspace(0.0, total_time, n_exact_steps + 1)

# --- Calcular pulsos (para plotting)
J_vals = np.array([J_ij_at_time(0, 1, t) for t in times_trot])
Bx_vals = np.zeros((N, len(times_trot)))
for idx, t in enumerate(times_trot):
    for i in range(N):
        Bvec = B_i_at_time(i, t)
        Bx_vals[i, idx] = Bvec[0]

# --- Evolución exacta (alta resolución) muestreada cada paso Trotter
psi_exact = initial_down_state(N)
M_t_exact = [np.mean(local_magnetizations_from_state(psi_exact))]  # lista iniciada en t=0
times_exact_sampled = [0.0]

for k in range(1, n_exact_steps + 1):
    t_mid = (k - 0.5) * dt_exact
    H_mid = build_H_matrix_time(t_mid)
    U = expm(-1j * H_mid * dt_exact)
    psi_exact = U @ psi_exact
    t_curr = k * dt_exact # nuevo
    M_t_exact.append(np.mean(local_magnetizations_from_state(psi_exact))) # nuevo
    times_exact_sampled.append(t_curr) # nuevo
    #if k % n_exact_substeps == 0:
    #    M_t_exact.append(np.mean(local_magnetizations_from_state(psi_exact)))

# --- Evolución trotter (coarse)
psi0 = initial_down_state(N)
M_t_trotter = [np.mean(local_magnetizations_from_state(psi0))]  # t=0

print("Comenzando bucle temporal: se llamará al qnode trotter para cada k y se integrará la evolución exacta por pasos")
for k in range(1, n_trotter_steps + 1):
    # Trotter: ejecutamos qnode con k pasos
    vals = trotter_evolution(k, scheme="first")
    mz_trot = np.mean([float(v) for v in vals])
    M_t_trotter.append(mz_trot)
    if (k % 10) == 0:
        print(f" step {k}/{n_trotter_steps}: t={(k*dt_trot):.3f}  M_trot={mz_trot:.4f}")

# --- Convertir listas a arrays (ahora que ya no usamos append)
M_t_exact = np.array(M_t_exact)
times_exact_sampled = np.array(times_exact_sampled)

M_t_trotter = np.array(M_t_trotter)

# -----------------------
# Graficar: pulsos vs magnetización comparación
# -----------------------
fig, (ax_pulses, ax_mag) = plt.subplots(2, 1, figsize=(10, 7), sharex=True,
                                         gridspec_kw={'height_ratios': [1, 1.1]})

# Pulsos: J(t)
ax_pulses.plot(times_trot, J_vals, label='J(t) (global)', color='tab:gray', linewidth=2)

# Pulsos: Bx por sitio
colors = plt.cm.viridis(np.linspace(0, 1, N))
for i in range(N):
    ax_pulses.plot(times_trot, Bx_vals[i], '--', label=f'Bx site {i}', color=colors[i])

ax_pulses.set_ylabel('Pulse amplitude')
ax_pulses.set_title('Pulsos en tiempo: J(t) y Bx^{(i)}(t)')
ax_pulses.legend(loc='upper right', fontsize='small', ncol=2)
ax_pulses.grid(True)

# Magnetización: comparar exact vs trotter
ax_mag.plot(times_exact_sampled, M_t_exact, '-', color='k', lw=1.5, label='exact (fine Δt, muestreada)')
ax_mag.plot(times_trot, M_t_trotter, 'o--', color='tab:red', label='trotter (coarse Δt)')
ax_mag.set_xlabel('t')
ax_mag.set_ylabel('M(t) = (1/N) Σ ⟨Z_j⟩')
ax_mag.set_title('Magnetización total (Exacta vs Trotter)')
ax_mag.grid(True)
ax_mag.legend()

plt.tight_layout()
plt.show()
```

```python
# trotter_braket_panel_left.py
import os
import numpy as np
import pennylane as qml
import matplotlib.pyplot as plt
from scipy.linalg import expm

# -----------------------
# parámetros del sistema (panel superior izquierdo: first-order, g=1)
# -----------------------
N = 5
J = 1.0
g = 1.0                      # primer panel: g = 1J (first-order)
dt = 0.2 / J                 # paso de Trotter (paper)
t_max = 2.0                  # eje J t en [0,2]
# Reducimos puntos temporales para minimizar llamadas remotas a Braket:
n_time = 9                   # reduce el número de tasks remotos (antes 21)
times = np.linspace(0.0, t_max, n_time)   # puntos Jt

# -----------------------
# Pauli matrices y utilidades
# -----------------------
sx = np.array([[0, 1], [1, 0]], dtype=complex)
sz = np.array([[1, 0], [0, -1]], dtype=complex)
si = np.eye(2, dtype=complex)

def kron_n(op_list):
    out = op_list[0]
    for op in op_list[1:]:
        out = np.kron(out, op)
    return out

def build_H_matrix(N, J, g):
    H = np.zeros((2**N, 2**N), dtype=complex)
    for j in range(N-1):
        ops = [si]*N
        ops[j]   = sz
        ops[j+1] = sz
        H += (-J) * kron_n(ops)
    for j in range(N):
        ops = [si]*N
        ops[j] = sx
        H += (-g) * kron_n(ops)
    return H

def initial_down_state(N):
    v = np.array([0,1], dtype=complex)   # |1> por convenio (spin down)
    psi = v
    for _ in range(N-1):
        psi = np.kron(psi, v)
    return psi

def local_magnetizations_from_state(psi, N):
    res = np.zeros(N, dtype=float)
    for j in range(N):
        ops = [si]*N
        ops[j] = sz
        Zj = kron_n(ops)
        res[j] = np.vdot(psi, Zj @ psi).real
    return res

# -----------------------
# Trotter step (primer orden) — mismo que antes
# -----------------------
def trotter_step_first_order(dt, J, g):
    # B_even (0-1, 2-3, ...)
    for j in range(0, N-1, 2):
        qml.IsingZZ(-2.0 * J * dt, wires=[j, j+1])
    # B_odd (1-2, 3-4, ...)
    for j in range(1, N-1, 2):
        qml.IsingZZ(-2.0 * J * dt, wires=[j, j+1])
    # A: campo X en cada qubit (full step)
    for j in range(N):
        qml.RX(-2.0 * g * dt, wires=j)

# -----------------------
# Configurar dispositivo Braket (usa tus valores)
# -----------------------
SV1_ARN = "arn:aws:braket:::device/quantum-simulator/amazon/sv1"

# Usa el bucket y prefijo que ya funcionan para ti (ejemplo del mensaje anterior)
s3_bucket = "amazon-braket-mi-programa-1053851512"   # <- usa el bucket que tienes
s3_prefix = "my-job"
s3_folder = (s3_bucket, s3_prefix)

# Número de shots para estimar expectativas en Braket. Más shots -> menor ruido por muestreo.
shots = 1024   # cámbialo a 256 o 512 si quieres ahorrar tiempo/coste

# Crear dispositivo remoto Braket para N qubits
dev = qml.device(
    "braket.aws.qubit",
    device_arn=SV1_ARN,
    s3_destination_folder=s3_folder,
    wires=N,
    shots=shots,
)

# -----------------------
# QNode remoto que implementa n_steps de Trotter (primer orden)
# -----------------------
@qml.qnode(dev)
def qnode_trotter_remote(n_steps: int, g_param: float):
    # preparar |↓↓...↓> = aplicar X en cada qubit sobre |0>
    for j in range(N):
        qml.PauliX(wires=j)
    # aplicar n_steps de Trotter (primer orden)
    for _ in range(n_steps):
        trotter_step_first_order(dt, J, g_param)
    # retornar expectativas <Z_j>
    return [qml.expval(qml.PauliZ(j)) for j in range(N)]

# -----------------------
# Ejecutar: evolución exacta (local) + ejecución remota (Braket)
# -----------------------
psi0 = initial_down_state(N)
H_exact = build_H_matrix(N, J, g)

M_exact = []
M_braket = []

print("ATENCIÓN: se enviarán tareas remotas a Braket — cada punto temporal crea una tarea.")
print(f"Se usarán {len(times)} puntos temporales, {shots} shots por tarea. Ajusta 'n_time' o 'shots' si quieres ahorrar tiempo/coste.\n")

for idx, t_Jt in enumerate(times):
    print(f"[{idx+1}/{len(times)}] Jt = {t_Jt:.3f} — preparando ejecución...")

    # tiempo físico t = (Jt)/J
    t_phys = t_Jt / J
    # exact evolution local
    U_t = expm(-1j * H_exact * t_phys)
    psi_t = U_t @ psi0
    local_exact = local_magnetizations_from_state(psi_t, N)
    M_exact.append(np.mean(local_exact))

    # número de pasos de Trotter (paper: dt = 0.2/J => n_steps = round(Jt/0.2))
    n_steps = int(np.round(t_Jt / 0.2))
    if n_steps < 0:
        n_steps = 0

    # Ejecutar QNode remoto — esto lanzará la tarea a Braket
    vals = qnode_trotter_remote(n_steps, g)
    # convertir a floats y calcular magnetización total
    M_braket.append(np.mean([float(v) for v in vals]))

    print(f"  -> n_steps={n_steps}, M_exact={M_exact[-1]:.4f}, M_braket={M_braket[-1]:.4f}")

# -----------------------
# Graficar
# -----------------------
plt.figure(figsize=(6,4))
plt.plot(times, M_exact, '-', color='k', lw=1.5, label='exact (local)')
plt.plot(times, M_braket, 'o-', label=f'Braket SV1 (first-order), shots={shots}')
plt.xlabel('J t')
plt.ylabel('M (magnetización total)')
plt.title('First-order (g=1J) — exact vs Braket (SV1)')
plt.ylim([-1.05, 1.05])
plt.grid(True)
plt.legend()
plt.show()
```
